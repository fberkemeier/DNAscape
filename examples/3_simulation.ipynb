# DNAscape simulation examples

This notebook focuses on `rsim`, the stochastic simulation interface. It starts from an inferred firing-rate profile and then simulates replication over the selected genomic interval.


## Setup


```python
%reload_ext autoreload
%autoreload 2

from dnascape import *
```

## Prepare a firing-rate profile


```python
chr_number = 1
output_resolution_kb = 1
interval_kb = (0, 60_000)
fork_speed = 1.4

replication_timing_full = loadcsv("example", example="replication_timing_hESC", chr_number=chr_number)
replication_timing, positions_kb = select_profile_region(
    replication_timing_full,
    output_resolution_kb=output_resolution_kb,
    interval_kb=interval_kb,
)

firing_rate = rmap(
    source={"replication_timing": replication_timing},
    target="firing_rate",
    assumptions={"fork_speed": fork_speed, "resolution": output_resolution_kb, "maxiter": 6},
)
```

## Run a stochastic simulation


```python
simulation = rsim(
    firing_rate=firing_rate,
    fork_speed=fork_speed,
    sim_number=200,
    resolution=output_resolution_kb,
    seed=1,
    time_statsQ=True,
    print_every=25,
)
```

```python
plotf(
    replication_timing,
    simulation["replication_timing"],
    x_array=positions_kb,
    labels=["Timing data", "Mean simulation"],
    invyQ=True,
    xtitle="Chromosome position (kb)",
    ytitle="Replication timing (min)",
)
```

```python
plotf(
    simulation["replicated_fraction"],
    simulation["active_forks"],
    labels=["Replicated fraction", "Active forks"],
    dual_axis=True,
    xtitle="Time (min)",
)
```

```python
plotf(
    simulation["origin_efficiency"],
    firing_rate,
    x_array=positions_kb,
    labels=["Origin efficiency", "Firing rate"],
    dual_axis=True,
    xtitle="Chromosome position (kb)",
)
```
