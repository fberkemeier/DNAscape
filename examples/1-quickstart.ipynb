# DNAscape quickstart

This notebook introduces the two core DNAscape workflows: `rmap` (mapping) and `rsim` (simulation).

`rmap` uses efficient mathematical relationships to map between experimental or model-derived DNA replication features, such as origin firing rates, fork speed, replicated fraction and replication timing (RT), under explicit mechanistic assumptions.

`rsim` simulates complete DNA replication programmes from origin firing and fork speed profiles, with flexible models of firing-factor availability and replication stress. It simulates initiation and fork progression throughout individual S phases, yielding features such as replication timing, inter-origin distances, origin efficiency and fork directionality.

The short examples below provide an introduction to DNAscape. Full details of both workflows, including how they can be combined, are provided in `2_mapping.ipynb` and `3_simulation.ipynb`.

## Load example data

We start from the hESC chromosome 1 replication-timing profile. The loaded profile is a one-dimensional array with one value per genomic bin, at a one kilobase resolution.


```python
replication_timing_full = loadcsv(
    "example",
    example="replication_timing_hESC",
    chr_number=1,
)
```

To focus on a specific genomic region, we can use `select_profile_region` to evaluate `replication_timing_full` at a new set of genomic positions, `positions_kb`, defined at a desired resolution. By default, we assume the same resolution as the full chromosome dataset.

```python
replication_timing, positions_kb = select_profile_region(
    replication_timing_full,
    interval_kb=(0, 60_000),
    output_resolution_kb=1,
)
```

A quick plot is useful before fitting. We may do that with `plotf`.

```python
plotf(
    replication_timing,
    x_array=positions_kb,
    invyQ=True,
    xtitle="Chromosome position (kb)",
    ytitle="Replication timing (min)",
)
```

With the timing dataset loaded, we may now use it to map and simulate DNA replication features.

## A. Mapping replication features

`rmap` is the mapping interface. It takes a source feature, a target feature, and a dictionary of assumptions. Internally, DNAscape selects the appropriate route for the requested mapping.


### Fit initiation rate from timing

Here we infer a space-dependent initiation-rate profile, $I(x)$, from replication timing, assuming a constant fork speed.

```python
fork_speed = 1.4  # kb/min

initiation_rate = rmap(
    source={
        "replication_timing": replication_timing,
        "fork_speed": fork_speed,
    },
    target="initiation_rate",
)

print(f"Fitted I(x) from replication timing data, assuming a constant fork speed of {fork_speed} kb/min.")
```

```python
plotf(
    initiation_rate,
    x_array=positions_kb,
    xtitle="Chromosome position (kb)",
    ytitle="Initiation rate",
)
```

### Map the fitted profile back to timing

Forward-mapping the fitted initiation rate profile back to timing gives a direct visual check of the mapping loop, this can later be compared with averaged stochastic simulations.


```python
replication_timing_mapped = rmap(
    source={
        "initiation_rate": initiation_rate,
        "fork_speed": fork_speed,
    },
    target="replication_timing",
)

plotf(
    replication_timing,
    replication_timing_mapped,
    x_array=positions_kb,
    labels=["Timing data (Repli-seq)", "Timing mapped from fitted initiation rate"],
    invyQ=True,
    xtitle="Chromosome position (kb)",
    ytitle="Replication timing (min)",
)
```

As expected, the fitting is not perfect, as the constant fork speed assumptions is too strong to describe the data. We may, for example, check a measure of the misfit by looking at their difference.

```python
plotf(
    np.abs(replication_timing - replication_timing_mapped),
    x_array=positions_kb,
    xtitle="Chromosome position (kb)",
    ytitle="Timing residual (min)",
)
```

### Mapping fork speed from replication timing and initiation rate

Conveniently, `rmap` also provides a tool to map fork speed along the chromosome, provided we have replication timing and origin firing data. By default, the mapping assumes fork speed varies only in space, but generalisations are possible (see `mapping.ipynb`).

```python
fork_speed_mapped = rmap(
    source={
        "initiation_rate": initiation_rate,
        "replication_timing": replication_timing,
    },
    target="fork_speed",
)

plotf(
    fork_speed_mapped,
    x_array=positions_kb,
    xtitle="Chromosome position (kb)",
    ytitle="Fork speed (kb/min)",
)
```

If we now forward-map the inferred initiation rate together with the space-dependent fork speed to replication timing, and compare the resulting misfit with that obtained under a constant fork speed assumption, we observe a substantial improvement.

```python
replication_timing_mapped_with_fork = rmap(
    source={
        "initiation_rate": initiation_rate,
        "fork_speed": fork_speed_mapped,
    },
    target="replication_timing",
)

plotf(
    np.abs(replication_timing - replication_timing_mapped),
    np.abs(replication_timing - replication_timing_mapped_with_fork),
    x_array=positions_kb,
    labels=["Misfit with initiation rate fitting", "Misfit with initiation rate and fork speed fitting"],
    xtitle="Chromosome position (kb)",
    ytitle="Timing residual (min)",
)
```

**Note.** DNAscape supports several modelling choices, but more constrained assumptions, such as constant fork speed, can provide a useful baseline, improve biological interpretability and reduce overfitting. Although allowing both $I(x,t)$ and $v(x,t)$ to vary in space and time can improve the fit to replication timing, the timing curve alone does not uniquely determine the pair $(I,v)$. Multiple initiation-rate and fork-speed profiles may therefore explain the same data, making additional assumptions or complementary datasets essential for meaningful inference. For illustration purposes, we keep the examples below for space-dependent initiation rate and constant fork speed.

### Map timing to fork directionality

Replication fork directionality (RFD) is another feature of the replication programme, and can be obtained directly from replication timing and fork speed.

```python
timing_reference = replication_timing_mapped

fork_directionality = rmap(
    source={
        "replication_timing": timing_reference,
        "fork_speed": fork_speed,
    },
    target="fork_directionality",
)

plotf(
    fork_directionality,
    x_array=positions_kb,
    xtitle="Chromosome position (kb)",
    ytitle="Fork directionality",
    ylims=(-1.05, 1.05),
)
```

## B. Simulating replication

In addition to mapping replication features under different assumptions, DNAscape can simulate replication directly. This provides a way to validate the mapping shortcuts introduced in the previous section, while also enabling the exploration of new scenarios and the testing of mechanistic hypotheses.

Replication simulations are performed using `rsim`, which generates replication dynamics from a supplied initiation-rate profile, fork speed and optional model assumptions, such as fork stalling, limiting firing factors or periodic genome geometry. We present a few examples below; see `3_simulation.ipynb` for a fuller overview of the available options and properties.

### Simulate from the fitted initiation rate

The simulation uses the fitted initiation-rate profile from section A. `rsim` prints progress by default.


```python
sim_number = 100

simulation = rsim(
    source={
        "initiation_rate": initiation_rate,
        "fork_speed": fork_speed,
    },
    assumptions={
        "sim_number": sim_number,
        "time_statsQ": True,
    },
)
```

```python
plotf(
    replication_timing,
    replication_timing_mapped,
    simulation["replication_timing"],
    x_array=positions_kb,
    labels=["Data (Repli-seq)", "Map (from initiation rate fitting)", f"Simulation (mean over {sim_number} runs)"],
    invyQ=True,
    xtitle="Chromosome position (kb)",
    ytitle="Replication timing (min)",
)
```

### Inspect simulation outputs

The simulation returns genome-wide profiles, time courses, and per-simulation summaries. These plots show the replicated fraction over time, origin efficiency along the genome, inter-origin distances, and compact summary statistics.


```python
plotf(
    simulation["replicated_fraction"],
    simulation["active_forks"],
    labels=["Replicated fraction", "Active forks"],
    dual_axis=True,
    xtitle="Time (min)",
    xlims=(0, 800),
)
```

```python
plotf(
    simulation["origin_efficiency"],
    initiation_rate,
    x_array=positions_kb,
    labels=["Origin efficiency", "Initiation rate"],
    dual_axis=True,
    xtitle="Chromosome position (kb)",
)
```

```python
ploth(
    simulation["inter_origin_distance"],
    xtitle="Inter-origin distance (kb)",
    xmax=1000,
)
```

### Combined genomic view

`mplotf` stacks several genomic tracks into one figure. This is often the most compact way to compare the input data, fitted maps, and simulation-derived profiles over the same interval.


```python
mplotf(
    [
        [replication_timing, replication_timing_mapped, simulation["replication_timing"]],
        initiation_rate,
        fork_directionality,
        [simulation["origin_efficiency"], initiation_rate],
    ],
    mlabels=[
        ["Timing data", "Deterministic map", "Mean simulation"],
        [None],
        [None],
        ["Origin efficiency", "Initiation rate"],
    ],
    mx_array=[positions_kb, positions_kb, positions_kb, positions_kb],
    mdual_axis=[False, False, False, True],
    minvyQ=[True, False, False, False],
    mylims=[(None, None), (None, None), (-1.05, 1.05), (None, None)],
    mxtitle=["", "", "", "Chromosome position (kb)"],
    mytitle=[
        "Replication timing (min)",
        "Initiation rate",
        "Fork directionality",
        "Simulation output",
    ],
    figsize=(10, 2.2),
    hspace=0.18,
)
```
