## Periodic replication example

This small bundled timing profile exercises the same mapping and simulation logic under periodic geometry.


```python
plasmid_timing = loadcsv("example", example="replication_timing_Pl", chr_number=1)
plasmid_timing = rescale(plasmid_timing, [0, 60])
plasmid_fork_speed = 0.6

plasmid_assumptions = {
    "resolution": 5,
    "geometry": "periodic",
    "maxiter": 6,
}

plasmid_firing_rate = rmap(
    source={
        "replication_timing": plasmid_timing,
        "fork_speed": plasmid_fork_speed,
    },
    target="firing_rate",
    assumptions=plasmid_assumptions,
)

plasmid_simulation = rsim(
    source={
        "firing_rate": plasmid_firing_rate,
        "fork_speed": plasmid_fork_speed,
    },
    assumptions={
        "sim_number": 50,
        "resolution": 5,
        "geometry": "periodic",
        "stall_rate": 0.1,
        "seed": 2,
        "time_statsQ": True,
        "print_every": 10,
    },
)
```

```python
plotf(
    plasmid_timing,
    plasmid_simulation["replication_timing"],
    labels=["Timing data", "Mean simulation"],
    invyQ=True,
    resolution=5,
    xtitle="Circular coordinate (kb)",
    ytitle="Replication timing (min)",
)
```
